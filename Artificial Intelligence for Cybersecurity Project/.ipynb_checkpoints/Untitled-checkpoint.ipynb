# Data Collection

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

file_path = 'src/Hulk-NoDefense.csv'
chunk_size = 10000  # Leggi 10.000 righe alla volta
chunks = []
# Usa un TextFileReader per leggere a chunk
for chunk in pd.read_csv(file_path, chunksize=chunk_size):
    chunks.append(chunk)

# Concatena tutti i chunk in un unico DataFrame
df_raw = pd.concat(chunks, ignore_index=True)
df_copy = df_raw.copy()
```

# Pre-processing

```python
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()

train_y = df_copy['Label']
train_y = label_encoder.fit_transform(train_y)  # Converti 'BENIGN', 'Hulk-NoDefense' in 0, 1

train_x = df_copy.drop(columns=['Label', 'Flow ID'])
```

Rimozione valori nulli

```python
rows_null = [486712, 702344, 837046]
for row in rows_null:
    train_x.loc[row, 'Flow Bytes/s'] = train_x.loc[row, 'Total Length of Fwd Packet']
    train_x.loc[row, 'Flow Packets/s'] = 1
    train_x.loc[row, 'Flow IAT Mean'] = 0
    train_x.loc[row, 'Flow IAT Std'] = 0
    train_x.loc[row, 'Flow IAT Max'] = 0
    train_x.loc[row, 'Flow IAT Min'] = 0
    
```

Gestione delle features categoriche

```python
import ipaddress

def convert_ip(ip):
    try:
        return int(ipaddress.ip_address(ip))
    except ValueError:
         return 0  # Sostituisci con NaN o un valore di default se preferisci

# Convertire Timestamp in variabili numeriche
train_x['Timestamp'] = pd.to_datetime(train_x['Timestamp'], format='%d/%m/%Y %I:%M:%S %p')
train_x['Hour'] = train_x['Timestamp'].dt.hour
train_x['Weekday'] = train_x['Timestamp'].dt.weekday
train_x['Month'] = train_x['Timestamp'].dt.month  # Aggiungi il mese, se desiderato

# Ora possiamo rimuovere la colonna originale 'Timestamp'
train_x=train_x.drop('Timestamp', axis=1)


# Preprocessing indirizzi IP
train_x['Dst IP'] = train_x['Dst IP'].apply(convert_ip)
train_x['Src IP'] = train_x['Src IP'].apply(convert_ip)

# Verifica il risultato
print(train_x.head())
```

## Solving unbalanced classes problem with an hybrid approach and smoothing the model with PCA and Feature Selection

```python
from imblearn.over_sampling import SMOTE
from sklearn.decomposition import PCA
from sklearn.feature_selection import SelectFromModel
from imblearn.under_sampling import RandomUnderSampler
from imblearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler

import warnings

# Sopprime i FutureWarnings
warnings.simplefilter(action='ignore', category=FutureWarning)

# Definizione di SMOTE e RUS
smote = SMOTE(sampling_strategy=0.5, random_state=42)
rus = RandomUnderSampler(sampling_strategy='majority', random_state=42)
scaler = StandardScaler()

# Creazione della pipeline con StandardScaler, SMOTE, RUS e Logistic Regression
pca_pipeline = Pipeline([
    ('scaler', scaler),
    ('smote', smote),
    ('rus', rus),
    ('pca', PCA(n_components=0.95)),                # PCA: Mantieni il 95% della varianza
    ('feature_selection', SelectFromModel(LogisticRegression(penalty='l1', solver='liblinear',random_state=42))),
    ('classifier', LogisticRegression(random_state=42, max_iter=1000))
])

# Definizione del K-Fold Cross Validator
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scoring = ['accuracy','precision','recall','f1']
# Cross-Validation con ritorno delle metriche su training e test set
scores = cross_validate(pca_pipeline,train_x,train_y,cv=skf, scoring=scoring,return_train_score=True)

# Stampa delle metriche per fold
print("Training Scores:")
for metric in scoring:
    print(f"{metric} (train): {scores['train_' + metric]}")

print("\nTest Scores:")
for metric in scoring:
    print(f"{metric} (test): {scores['test_' + metric]}")

metrics_dict = {
    "Metric": scoring,
    "Train Scores": [scores['train_' + metric].mean() for metric in scoring],
    "Test Scores": [scores['test_' + metric].mean() for metric in scoring],
    "Differenza Media (Train - Test)": [scores['train_' + metric].mean() - scores['test_' + metric].mean() for metric in scoring]
}

# Creazione del DataFrame per una visualizzazione tabellare
metrics_df = pd.DataFrame(metrics_dict)

# Stampa della tabella
print(metrics_df.to_string(index=False))
```

## Smote and Rus effects on first fold

```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import StratifiedKFold
from imblearn.over_sampling import SMOTE
from imblearn.under_sampling import RandomUnderSampler

# Definizione di SMOTE e RUS
smote = SMOTE(sampling_strategy=0.5, random_state=42)
rus = RandomUnderSampler(sampling_strategy='majority', random_state=42)

# Creazione dei fold
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Estrazione e trasformazione per StratifiedKFold
train_y = pd.Series(train_y)
train_idx, test_idx = next(skf.split(train_x, train_y))
x_train, y_train = train_x.iloc[train_idx], train_y.iloc[train_idx]

# Conta delle classi prima dell'applicazione di SMOTE e RUS
class_counts_original_skf = y_train.value_counts()
total_instances_original_skf = class_counts_original_skf.sum()

x_scaled = StandardScaler().fit_transform(x_train)
# Applicazione di SMOTE sul primo fold con skf
x_resampled_skf, y_resampled_skf = smote.fit_resample(x_scaled, y_train)

# Conta delle classi dopo SMOTE
class_counts_smote_skf = y_resampled_skf.value_counts()
total_instances_smote_skf = class_counts_smote_skf.sum()

# Applicazione di RUS sul dataset già bilanciato con SMOTE
x_resampled_skf, y_resampled_skf = rus.fit_resample(x_resampled_skf, y_resampled_skf)

# Conta delle classi dopo RUS
class_counts_rus_skf = y_resampled_skf.value_counts()
total_instances_rus_skf = class_counts_rus_skf.sum()


# Creazione dei subplot
fig, axes = plt.subplots(1, 3, figsize=(18, 10))

# Grafico per StratifiedKFold
axes[0].pie(class_counts_original_skf, labels=class_counts_original_skf.index, autopct='%1.1f%%', startangle=90, colors=['lightblue', 'orange'])
axes[0].set_title(f"Distribuzione originale (skf)\nTotale: {total_instances_original_skf} istanze")

axes[1].pie(class_counts_smote_skf, labels=class_counts_smote_skf.index, autopct='%1.1f%%', startangle=90, colors=['lightblue', 'orange'])
axes[1].set_title(f"Distribuzione dopo SMOTE (skf)\nTotale: {total_instances_smote_skf} istanze")

axes[2].pie(class_counts_rus_skf, labels=class_counts_rus_skf.index, autopct='%1.1f%%', startangle=90, colors=['lightblue', 'orange'])
axes[2].set_title(f"Distribuzione dopo RUS (skf)\nTotale: {total_instances_rus_skf} istanze")

# Mostra i grafici
plt.tight_layout()
plt.show()
```

## PCA result on first fold

```python

# Esegui PCA sui dati
pca = PCA(n_components=0.95)
X_pca = pca.fit_transform(x_resampled_skf)

# Numerico: Varianza spiegata
explained_variance = pca.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

# Numero di componenti principali selezionati
n_components = pca.n_components_

# Numero di feature originali
original_feature_count = x_resampled_skf.shape[1]

# Stampa i dettagli
print(f"Numero di feature originali: {original_feature_count}")
print(f"Numero di componenti principali selezionati: {n_components}")
print(f"Varianza cumulativa spiegata: {cumulative_variance[-1]:.2f}")

# Grafico della Varianza Spiegata
plt.figure(figsize=(12, 6))

# Varianza spiegata individuale
plt.bar(range(1, len(explained_variance) + 1), explained_variance, alpha=0.7, align='center')
plt.step(range(1, len(cumulative_variance) + 1), cumulative_variance, where='mid', label='Varianza Cumulativa')

# Titolo con numero di feature originali
plt.title(f'Varianza Spiegata per Componente (Feature Originali: {original_feature_count})')

# Etichette degli assi
plt.xlabel('Componente Principale')
plt.ylabel('Proporzione di Varianza Spiegata')

# Legenda
plt.legend()

plt.tight_layout()
plt.show()
```

## Learning Curve

```python
import matplotlib.pyplot as plt
from sklearn.model_selection import learning_curve
import warnings
import numpy as np

# Sopprime i FutureWarnings
warnings.simplefilter(action='ignore', category=FutureWarning)

# Funzione per plottare learning curves per diverse metriche
def plot_learning_curve(estimator, X, y, metrics, cv, train_sizes=np.linspace(0.3, 1.0, 10)):
    # Creiamo la figura e la griglia di 2x2
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))
    axes = axes.ravel()  # Flatten l'array 2x2 in un array 1D per facilitarne l'iterazione
    
    # Ciclo per ciascuna metrica e plottiamo la learning curve
    for i, metric in enumerate(metrics):
        # Calcola le learning curves
        train_sizes, train_scores, val_scores = learning_curve(
            estimator, X, y, cv=cv, scoring=metric, n_jobs=-1, train_sizes=train_sizes
        )
        
        # Calcola le medie e le deviazioni standard
        train_mean = np.mean(train_scores, axis=1)
        val_mean = np.mean(val_scores, axis=1)
        train_std = np.std(train_scores, axis=1)
        val_std = np.std(val_scores, axis=1)

        # Plot delle learning curves
        axes[i].plot(train_sizes, train_mean, label='Training score', color='b')
        axes[i].plot(train_sizes, val_mean, label='Validation score', color='r')
        axes[i].fill_between(train_sizes, train_mean - train_std, train_mean + train_std, alpha=0.2, color='b')
        axes[i].fill_between(train_sizes, val_mean - val_std, val_mean + val_std, alpha=0.2, color='r')

        # Aggiungi titoli, etichette e legende
        axes[i].set_title(f'Learning Curve ({metric})')
        axes[i].set_xlabel('Training Set Size')
        axes[i].set_ylabel(metric.capitalize())
        axes[i].legend(loc='best')
        axes[i].grid()

    plt.tight_layout()  # Per migliorare l'aspetto dei subplot
    plt.show()

# Esempio: Genera learning curves per diverse metriche
metrics = ['accuracy', 'precision', 'recall', 'f1']
plot_learning_curve(pca_pipeline, train_x, train_y, metrics=metrics, cv=skf)
```

## Roc Auc Curve

```python
from sklearn.metrics import roc_curve, auc
# Calcolo e tracciamento della curva ROC e AUC
# Addestra il modello sull'intero training set
pca_pipeline.fit(train_x, train_y)

# Ottieni le probabilità di previsione per la classe positiva
y_prob = pca_pipeline.predict_proba(train_x)[:, 1]

# Calcola la curva ROC
fpr, tpr, thresholds = roc_curve(train_y, y_prob)
roc_auc = auc(fpr, tpr)

# Traccia la curva ROC
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic Curve')
plt.legend(loc='lower right')
plt.show()
```

## Residuals Plot

```python
from sklearn.metrics import mean_squared_error

# Inizializza la lista per raccogliere i residui e le probabilità previste
residuals = []
predicted_probs = []
true_labels = []
train_y = pd.Series(train_y)

# Definizione del StratifiedKFold
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Ciclo sui fold di cross-validation
for train_index, val_index in skf.split(train_x, train_y):
    # Usa .iloc per selezionare i dati di addestramento e validazione
    X_train, X_val = train_x.iloc[train_index], train_x.iloc[val_index]
    y_train, y_val = train_y.iloc[train_index], train_y.iloc[val_index]
     
    # Fit del modello sui dati di addestramento
    pca_pipeline.fit(X_train, y_train)
    
    # Predizione delle probabilità per il fold di validazione
    y_pred_prob = pca_pipeline.predict_proba(X_val)  # Probabilità predette
    
    # Scegliamo le probabilità per la classe positiva (ad esempio, la classe 1)
    y_pred_class = y_pred_prob[:, 1]  # Probabilità previste per la classe positiva (1)
    
    # Calcolo dei residui (probabilità previste - etichette vere)
    residuals_fold = y_pred_class - y_val
    
    # Aggiungi ai risultati
    residuals.extend(residuals_fold)
    predicted_probs.extend(y_pred_class)
    true_labels.extend(y_val)

# Creiamo il grafico dei residui
plt.figure(figsize=(8, 6))
plt.scatter(predicted_probs, residuals, color='blue', alpha=0.5)
plt.axhline(y=0, color='r', linestyle='--', label='Zero Residual Line')
plt.title("Residual Plot")
plt.xlabel("Predicted Probability of Class 1")
plt.ylabel("Residuals")
plt.legend()
plt.grid(True)
plt.show()
```
